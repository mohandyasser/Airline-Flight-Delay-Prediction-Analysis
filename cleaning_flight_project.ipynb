#Cleaning Data

FL_DATE

```python
 # Remove 0:00 from the date
df["FL_DATE"] = pd.to_datetime(df["FL_DATE"]).dt.date
df["FL_DATE"] = pd.to_datetime(df["FL_DATE"])
df["FL_DATE"]
```

TAIL_NUM

```python
df["TAIL_NUM"] = df["TAIL_NUM"].fillna("Unknown")
df.isnull().sum()
```

```python
# Keep city name only
df["DEST_CITY_NAME"] = (
    df["DEST_CITY_NAME"]
    .astype(str)
    .str.split(",")
    .str[0]
    .str.strip()
)
```

```python
df["DEST_CITY_NAME"]
```

DEP_DELAY

```python
df.info()
```

```python
#Replace missing values with Median
df["DEP_DELAY"] = df["DEP_DELAY"].fillna(df["DEP_DELAY"].median())
df['DEP_DELAY'] = df['DEP_DELAY'].astype('int64')
df.info()
```

 ARR_DELAY

```python
# Replace missing values with Median
df["ARR_DELAY"] = df["ARR_DELAY"].fillna(df["ARR_DELAY"].median())
df['ARR_DELAY'] = df['ARR_DELAY'].astype('int64')
df.info()
```

```python
pd.value_counts(df["ARR_DEL15"])
```

TAXI_OUT

```python
# Replace missing values with Mean
df["TAXI_OUT"] = df["TAXI_OUT"].fillna(df["TAXI_OUT"].mean())
df.isnull().sum()
```

CANCELLATION_CODE

```python
# Drop columns
df = df.drop(
    columns=["CANCELLATION_CODE"]
)
df.columns
```

ACTUAL_ELAPSED_TIME

```python
# Replace missing values with Median
df["ACTUAL_ELAPSED_TIME"] = df["ACTUAL_ELAPSED_TIME"].fillna(df["ACTUAL_ELAPSED_TIME"].median())
df['ACTUAL_ELAPSED_TIME'] = df['ACTUAL_ELAPSED_TIME'].astype('int64')
df.info()
```

CRS_DEP_TIME

```python
# Convert CRS_DEP_TIME to HH:MM
def convert_time(value):
    if pd.isna(value):
        return np.nan

    value = int(value)

    hour = value // 100
    minute = value % 100

    return f"{hour:02d}:{minute:02d}"


df["CRS_DEP_TIME"] = df["CRS_DEP_TIME"].apply(convert_time)
```

```python
df["CRS_DEP_TIME"]
```

CRS_ARR_TIME

```python
# 13. Convert CRS_ARR_TIME to HH:MM
df["CRS_ARR_TIME"] = df["CRS_ARR_TIME"].apply(convert_time)
```

```python
df["CRS_ARR_TIME"]
```

ARR_TIME_BLK

```python
# Convert ARR_TIME_BLK
# Example: 2200-2259 -> 22:00-22:59
def convert_time_block(value):

    if pd.isna(value):
        return np.nan

    value = str(value)

    start, end = value.split("-")

    start_time = convert_time(int(start))
    end_time = convert_time(int(end))

    return f"{start_time}-{end_time}"


df["ARR_TIME_BLK"] = df["ARR_TIME_BLK"].apply(
    convert_time_block
)
```

```python
df["ARR_TIME_BLK"]
```
